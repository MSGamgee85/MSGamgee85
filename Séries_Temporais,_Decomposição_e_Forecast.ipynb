## **Séries Temporais**

---


A maioria dos problemas em finanças e economia envolve o estudo de **sistemas dinâmicos** — fenômenos e eventos que mudam com o passar do tempo.




Portanto, torna-se necessário o uso de **ferramentas matemáticas** que permitem analisar o <u>comportamento e a relação</u> entre os dados (informações) para responder a questão _"por que aconteceu?"_ e obter estimativas (previsões) para responder a pergunta _"o que pode acontecer?"_ e, assim, facilitar o processo da **tomada de decisão**.  


<br>

---

<u>Definição</u>: Uma **série temporal (st)**, denotada por {$ \sf y_t $}, pode ser entendida como uma **conjunto de observações** $\sf (y_o, y_1, y_2, ..., y_n,...)$ indexados pela data (tempo) de cada observação.

Geralmente estamos interessados em coletar uma amostra (conjunto de dados) a partir de uma data inicial (por exemplo, $\sf t = 1$) e terminando em outra (digamos, $\sf t = T$). Essa amostra de dados $\sf (y_1, y_2, ..., y_T)$ pode ser vista como um segmento finito de uma **sequência infinita**, denotada por:

{$\sf y_t$}$_{\sf t=-\infty}^{\infty}= \, $ {$ \, \sf ..., y_{-1}, y_o, y_1, y_2, ...., y_T, y_{T+1}, y_{T+2}, ...$}


---

## **Modelo ARIMA**

---
Existem diversos modelos de st, representadas por funções que descrevem com maior ou menor **precisão** o <u>comportamento estatístico</u> de um conjunto de dados.


O **ARIMA** (_AutoRegressive Integrated Moving Average_ _ média móvel autorregressiva integrada numa tradução livre) é um modelo matemático para **séries temporais não-estacionárias** [i.e., seu valor esperado $\mu$ não é constante: $\mu = \mu(t)$] e que possuem **variação periódica** (oscilações de valor ao longo do tempo).


---

O acrônimo ARIMA é uma **fusão** de <u>3 componentes</u>:

* **AR = autoregressive**; indica que a variável de interesse é regredida a seus valores anteriores:

$ \qquad \boxed{\sf y_t = \theta_o + \theta_1 y_{t-1} + \theta_2 y_{t-2}+...}$

* **MA = moving average**; denota que o erro de regressão é uma combinação linear dos erros que ocorreram no instante de tempo atual e no passado:

$ \qquad \boxed{\sf \varepsilon_t = \phi_o + \phi_1 \varepsilon_{t-1} + \phi_2 \varepsilon_{t-2}+...}$


* **I = integrated**; significa que cada valor da série foi substituído pela diferença entre o valor atual e o anterior:

$ \qquad \boxed{\sf y_t \; \rightarrow \, \Delta y_t = y_t - y_{t-1}}$

---

Seja um série temporal {$\sf y_t$} onde t é número inteiro e $\sf y_t$ são números reais, um modelo ARMA(p, q) é dado por:

$\boxed{\sf y_t - \theta_1 y_{t-1} - ... - \theta_p y_{t-p} = \varepsilon_t + \phi_1 \varepsilon_{t-1}+...+\theta_q \varepsilon_{t-q}}$

ou analogamente:

$\boxed{\sf \bigg ( 1- \displaystyle \sum_{i=1}^p \theta_i L^i \bigg )y_t = \bigg ( 1 + \displaystyle \sum_{i=1}^q \phi_i L^i \bigg ) \varepsilon_t}$

onde:

* $L^i$ = operador "atraso" (_lag_ em inglês).



## **Forecast**

---

Em muitas situações estamos interessados em fazer uma estimativa dos valores futuros de uma série temporal; chamamos de _forecast_ (previsão) essa estimativa, baseada em uma amostra de dados da série.

Suponha que queremos prever o valor de uma variável $\sf Y_{t+1}$ baseado no conjunto de variáveis $\sf X_t$ observadas na data t. Por exemplo, desejamos estimar o valor futuro de $\sf Y_{t+1}$ a partir dos seus n mais recente valores. Neste caso, $\sf X_t$ seria formado por uma constante mais $\sf Y_t, Y_{t+1}, ...,$ e $\sf Y_{t-n+1}$

<u>Definição</u>: $\sf Y^*_{t+1|t}$ = previsão (forecast) de $\sf Y_{t+1}$ baseada em $\sf X_t$

<br>


### **Métricas de Desempenho do Forecast**

---



Para avaliar a utilidade/desempenho de uma previsão, nós precisamos especificar uma **função de perda (loss function)** ou um resumo de quão preocupados estamos que nossa estimativa está distante de um determinado valor.


<br>

O **erro absoluto médio** (_Mean Absolute Error_) mede a diferença entre o valor previsto e o observado e é calculado da seguinte forma:

$\boxed{\sf MAE = \dfrac{1}{n} \displaystyle \sum_{i=1}^n \bigg |Y_{t+1} - Y^*_{t+1|t} \bigg |}$

onde:

* $\sf n$ = número de observações da amostra;
* $\sf Y_{t+1}$ = valor observado;
* $\sf Y^*_{t+1|t}$ = valor da previsão (_forecast_).

<br>

Resultados mais convenientes, contudo, são obtidos quando assumimos uma **função quadrática de perda**. Uma função quadrática de perda significa escolher a previsão $\sf Y^*_{t+1|t}$ de tal forma a minimizar

$\sf E[(Y_{t+1} - Y^*_{t+1|t})^2]$

A expressão acima é conhecida como **erro quadrático médio** (_Mean Squared Error_) associado a previsão $\sf Y^*_{t+1|t}$ denotado por:

$\boxed{\sf MSE(Y^*_{t+1|t}) ≡ E[(Y_{t+1} - Y^*_{t+1|t})^2]}$

A previsão com o menor erro quadrático médio é o valor esperado de $\sf Y_{t+1}$ condicionado a $\sf X_t$:

$\sf Y^*_{t+1|t} = E(Y_{t+1}|X_t)$

<br>

Frequentemente é interessante apurar o valor da função erro na <u>mesma unidade de medida</u> dos dados da série original; neste caso usamos a **raiz do erro quadrático médio** (_Root Mean Squared Error_):

$\sf RMSE = \sqrt{MSE}$

$\boxed{\sf RMSE = \sqrt{E[(Y_{t+1} - Y^*_{t+1|t})^2]}}$

### **Função (Parcial) de Autocorrelação - ACF e PACF**

---

Em análise de séries temporais é comum querermos saber a **correlação** entre uma dada observação e seus valores passados. Com isso, é possível identificar comportamentos de tendência, sazonalidade e periodicidade.

Portanto, é necessário conhecer duas ferramentas matemáticas que fazem esse trabalho: a **função de autocorrelação** (_AutoCorrelation Function - ACF_) e a **função parcial de autocorrelação** (_Partial AutoCorrelation Function - PACF_).

<br>

* **ACF** mede a correlação entre um dado ponto e seus valores anteriores, considerando todos os lags (atrasos) intermediários. Ele fornece uma visão abrangente como cada observação se relaciona com seus valores passados.

<br>

* **PACF**, por sua vez, isola a direta correlação entre um dado ponto e um lag específico, enquanto controle a influência de outros lags. Fornece, portanto, um visão mais focada da relação entre um determinado ponto e seu passado próximo.

---

**Figura 4.** Cada pico do gráfico representa o número de atrasos (lags) e o valor da PACF. Valores positivos e negativos indicam correlações  positivas ou negativas, respectivamente enquanto valores próximos de zero sugerem fraca correlação para aquela lag. Um pico significativo em um lag particular indica forte correlação entre a variável e seu valor naquele lag, independente dos demais lags. Por exemplo, um PACF que mostra um pico significativo na lag 1 mas nenhum outro pico significativo nos lags subsequentes sugere um processo autorregressivo de primeira ordem AR(1).

```python
# -- Bibliotecas utilizadas --
import numpy as np                                      # funções matemáticas e operações com arrays (vetores e matrizes)
import pandas as pd                                     # estruturas tabulares de análise e organização de dados
import matplotlib.pyplot as plt                         # geração e visualização de gráficos
import pmdarima as pm                                   # séries temporais usando o framework ARIMA
from statsmodels.tsa.seasonal import seasonal_decompose # função do statsmodels de decomposição sazonal
from statsmodels.graphics.tsaplots import plot_acf      # importação função de autocorrelação
from statsmodels.graphics.tsaplots import plot_pacf     # importação função de autocorrelação
from sklearn.metrics import mean_absolute_error         # função erro absoluto médio (MAE) do statsmodels
from sklearn.metrics import mean_squared_error          # função erro quadrático médio (MSE) do statsmodels
```

## **1. Construindo uma Série Temporal**

---

```python
# -- Gera uma série temporal a partir de suas componentes --

meses = 120                                         # total de meses (pontos) da série [10 anos]

datas = pd.date_range("2015-01-01",                 # cria as datas mensais (eixo horizontal)
                      periods=meses, freq="ME")

tend = np.linspace(50, 120, meses)                  # distribui 120 pontos igualmente espaçados no intervalo y = [50 a 120]
sazonal = 10 * np.sin(2 * np.pi * datas.month / 12) # padrão que se repete a cada 12 meses [1 ano] = a*sen(2pi*t/12)
ruido = np.random.normal(0, 5, meses)               # gera pontos aleatórios usando distro normal (mu=0, sigma=5, qtd de pontos)

st = tend + sazonal + ruido                         # equação da série
```

```python
# -- Criação do DataFrame para organizar os dados da st --

dados = pd.DataFrame({"Data": datas,   # armazena os dados da st (data e valor) em uma df de duas colunas
                     "Valor": st})
dados.set_index("Data", inplace=True)  # define a coluna de datas como o índice da tabela
dados                                  # exibe as 5 primeiras e 5 últimas linhas
```

```python
# -- Gera a figura e gráfico (curva) da série --

plt.figure(figsize=(12,5))          # cria uma figura de 12 x 5 pols
plt.plot(dados.Valor)               # gera a curva da série temporal [df.coluna]
plt.title("Série Temporal")         # título do gráfico
plt.xlabel("Tempo")                 # título do eixo x (horizontal)
plt.ylabel("Valor")                 # título do eixo y (vertical)
plt.grid(alpha=0.5)                 # adiciona grade com transparência de 50%
plt.show                            # mostra a figura
```

Note que foi criado uma série temporal com tendência crescente, sazonalidade que se repete anualmente (12 meses) e uma componente de ruído que provoca uma oscilação aleatória dos pontos.  

## **2. Decomposição de uma ST**

---

```python
# -- Interpolação dos dados (necessária para a decomposição e forecast) --
dados.Valor = dados.Valor.interpolate() # função de interpolação [linear - padrão Pandas]
dados.Valor                            # imprime o cabeçalho e o rodapé do df (10 linhas)
```

**Observação**: A função de interpolação retornou os mesmos valores pois construímos uma st sem "buracos" (falhas de medição) o que deixou o resultado redundante; em um caso real, contudo, essa etapa é realmente necessária para a decomposição e o forecast.

```python
# Identifica e limpa 'outliers' (valores discrepantes ou pontos fora da curva)
z = np.abs((dados.Valor - dados.Valor.mean())/ dados.Valor.std())            # calcula o z-score z = |x - μ|/|σ|
dados["Valor_ajustado"] = np.where(z > 3, dados.Valor.median(), dados.Valor) # aloca os valores calculados na df original com a nova coluna "Valor_ajustado"
dados = dados[["Valor", "Valor_ajustado"]]
dados
```

```python
# Decomposição da série utilizando a coluna "Valor_ajustado"
decomp = seasonal_decompose(dados.Valor_ajustado.dropna(), model = "additive", period = 12)    # função que decompõe a ts
decomp.plot()                                                                                  # gera os gráficos das componentes
plt.show()                                                                                     # mostra a figura
```

## **3. Forecast (previsão) de uma ST**

---

```python
# Dividindo os dados em treino e teste (usando 100 meses para treino e 20 para teste)
train = dados["Valor_ajustado"].iloc[:100]
test = dados["Valor_ajustado"].iloc[100:]

# Ajustando a função auto_arima de forma mais simples e direta
model = pm.auto_arima(
    train,
    seasonal=True,
    m=12,
    trace=True,
    error_action="ignore",
    suppress_warnings=True
)
print(model.summary())
```

```python
# Previsão do modelo
forecast = model.predict(n_periods=len(test))
forecast = pd.Series(forecast, index=test.index)
```

```python
# Geração da figura e do gráfico da ts e da previsão
plt.figure(figsize=(12,5))
plt.plot(train, label='Dados de Treino')
plt.plot(test, label = 'Valores Reais', linewidth = 2)
plt.plot(forecast, label = 'Valores Previsão', linestyle = '--')
plt.title("Previsão vs Real")
plt.xlabel("Tempo")
plt.ylabel("Valor")
plt.legend()
plt.grid(alpha=0.5)
plt.show()
```

```python
# -- Cálculo das métricas de erro do modelo --

mae = mean_absolute_error(test, forecast) # erro absoluto médio (MAE)
mse = mean_squared_error(test, forecast)  # erro quadrático médio (MSE)
rmse = np.sqrt(mse)                       # raiz do erro quadrático médio (RMSE)

print("--- Métricas de Avaliação --- ")
print("=" * 30)
print(f"MAE: {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
```

**Observações**:

* **MAE (Erro Absoluto Médio)** = `5.0007`: Em média, as previsões do modelo erram por aproximadamente 5.00 unidades para cima ou para baixo em relação aos valores reais. Como a nossa série oscila em torno de valores entre 100 e 120 no conjunto de teste, um erro médio de 5 unidades representa uma ótima precisão (cerca de 4% a 5% de erro relativo).

* **RMSE (Raiz do Erro Quadrático Médio)** = `6.1994`: Esta métrica penaliza erros de maior magnitude de forma mais severa. O fato de o RMSE (6.20) ser próximo ao MAE (5.00) é um excelente sinal, pois indica que o modelo possui previsões consistentes e está livre de grandes desvios ou erros discrepantes (outliers de previsão).


```python

# Obtendo os resíduos do modelo no conjunto de treino
residuos = model.resid()

# Criando a figura para os gráficos de ACF e PACF
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Gráfico de ACF (Autocorrelação)
plot_acf(residuos, lags=24, ax=axes[0])
axes[0].set_title("Autocorrelação (ACF) dos Resíduos")
axes[0].grid(alpha=0.3)

# Gráfico de PACF (Autocorrelação Parcial)
plot_pacf(residuos, lags=24, ax=axes[1])
axes[1].set_title("Autocorrelação Parcial (PACF) dos Resíduos")
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
