## **1. Introdução**
---

<u>**Situação**</u>: `Disputa comercial entre dois países fictícios, Tropicália e La Plata`


Dois países fictícios, **Tropicália** e **La Plata**, disputam o mercado consumidor da Europa. Tropicália tem a opção de assinar um acordo comercial vantajoso com a União Europeia. No entanto, La Plata já declarou que, caso isso ocorra, entrará em guerra para tentar impedir a perda de competitividade. Esse cenário representa um dilema estratégico de tomada de decisão internacional com implicações econômicas e políticas.

A motivação para esse trabalho são as tensões geopolíticas vivenciadas no mundo atual, com o acirramento das disputas multilaterais pelo comércio global e a tentativa de reafirmação do poder e controle hegemônico dos EUA através do uso da força militar e econômica.

## **2. Formulação do problema de tomada de decisão**
---

**Jogadores**:

  * País Tropicália

  * País La Plata

**Estratégias**:

* Tropicália: {Assinar acordo, Não assinar}

* La Plata: {Entrar em guerra, Não entrar em guerra}

**Hipóteses**:

* Jogo simultâneo (ambos decidem sem conhecer a escolha do outro).

* Informação completa (cada país conhece as possíveis estratégias e consequências).

* Informação imperfeita (não sabem a decisão do outro no momento da escolha).

## **3. Solução do problema de decisão**
---

Definimos os **payoffs** <u> (valores hipotéticos): </u>

* Se Tropicália assina e La Plata não entra em guerra → Tropicália ganha +5 (acordo vantajoso), La Plata perde -5.

* Se Tropicália assina e La Plata entra em guerra → Tropicália perde -10 (custos da guerra), La Plata perde -8 (custos da guerra).

* Se Tropicália não assina e La Plata não entra em guerra → ambos ficam neutros (0,0).

* Se Tropicália não assina e La Plata entra em guerra → Tropicália perde -3 (instabilidade), La Plata perde -2 (custos militares sem benefício).

<u> **Tabela de Payoffs** </u>

| Tropicália/La Plata | Guerra (-) | Paz (+) |
|:-------------------:|:----------:|:-------:|
| **Assinar acordo**  | (-10,-8)   | (5,-5)  |
| **Não assinar**     | (-3,-2)    | (0,0)   |


### <u> **Conceito de solução** </u>

O conceito de solução escolhido é o Equilíbrio de Nash: cada jogador escolhe a melhor resposta dado o comportamento do outro.

* Se Tropicália assina, La Plata prefere guerra (-8 > -5).

* Se Tropicália não assina, La Plata prefere paz (0 > -2).

* Tropicália, sabendo disso, avalia:

  * Se assinar → resultado esperado (-10).

  * Se não assinar → resultado esperado (0).

**Equilíbrio de Nash**: Tropicália **não assina** e La Plata **não entra em guerra** → (0,0).

## **Simulações do jogo**
---

```python
# -- Bibliotecas utilizadas --
import sys
!{sys.executable} -m pip install nashpy # instala pacote de teoria de jogos
import nashpy as nash # instala pacote nashpy
import numpy as np  # operações matemáticas e cálculos com arrays
import matplotlib.pyplot as plt # geração e visualização de gráficos
import seaborn as sns # geração e visualização de heatmaps
```

```python
#-- Simulação do jogo --

# Payoffs para Tropicália (linha) e La Plata (coluna)
# Matriz de Tropicália
A = np.array([[-10, 5],
              [-3, 0]])

# Matriz de La Plata
B = np.array([[-8, -5],
              [-2, 0]])

# Criando o jogo
game = nash.Game(A, B)

print("Jogo entre Tropicália e La Plata")
print("=" * 32)
print("Matriz de Tropicália:\n", A)
print("Matriz de La Plata:\n", B)

# Encontrando equilíbrios de Nash
equilibrios = list(game.support_enumeration())
print("\nEquilíbrios de Nash encontrados:")
for eq in equilibrios:
    print(eq)
```

```python
# --- Heatmaps ---
payoffs_tropic = [[-10, 5], [-3, 0]]
payoffs_laplata = [[-8, -5], [-2, 0]]

fig, axes = plt.subplots(1, 2, figsize=(10,4))

sns.heatmap(payoffs_tropic, annot=True, cmap="RdYlGn", center=0, ax=axes[0])
axes[0].set_title("Payoffs Tropicália")
axes[0].set_xticklabels(["Guerra", "Paz"])
axes[0].set_yticklabels(["Assinar", "Não assinar"], rotation=0)

sns.heatmap(payoffs_laplata, annot=True, cmap="RdYlGn", center=0, ax=axes[1])
axes[1].set_title("Payoffs La Plata")
axes[1].set_xticklabels(["Guerra", "Paz"])
axes[1].set_yticklabels(["Assinar", "Não assinar"], rotation=0)

plt.tight_layout()
plt.show()
```

Foram gerados dois gráficos lado a lado:

1. <u>Heatmap de Tropicália:</u> mostra os ganhos/perdas em cada cenário.

      * Verde indica ganhos (máximo +5).

      * Vermelho indica perdas (mínimo -10).

2. <u>Heatmap de La Plata:</u> mostra os resultados para La Plata.

      * Verde indica neutralidade (0).

      * Vermelho indica perdas (até -8).

Esses mapas permitem visualizar rapidamente que a guerra é sempre ruim para La Plata, e que Tropicália só teria ganhos se La Plata não reagisse com guerra (o que seria pouco provável).

## **4. Discusão dos resultados**
---

O modelo matemático mostra que, apesar do ganho potencial do acordo, o risco da guerra torna a decisão inviável para Tropicália. O equilíbrio de Nash revela que ambos os países preferem evitar o conflito, mesmo que isso implique abrir mão de vantagens econômicas.

Os *heatmaps* obtidos os reforçam nossa intuição: o acordo é arriscado demais para Tropicália, pois a resposta provável de La Plata é a guerra. Portanto, o **equilíbrio de Nash (0,0)** aparece como a opção menos prejudicial para ambos. Esse resultado reflete a lógica da teoria dos jogos aplicada à geopolítica e finanças corporativas: decisões estratégicas não dependem apenas de ganhos imediatos, mas também da reação (estratégia) do adversário.

Para completar, no mundo real as relações entre os países tendem a ser multilaterais, com interesses complexos que refletem o grau de relacionamento entre as partes e cujos ganhos e perdas que vão além do jogo simulado neste trabalho.
